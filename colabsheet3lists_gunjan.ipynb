```python
# Let's use append to append an element to our list.

subjects.append('BE')
print(subjects)
```

```python
# To insert an element somewhere in the middle, say 3rd index

subjects.insert(3, 'ME')           # i.e. insert requires position and value
print(subjects)
```

```python
# You can append another variable as well

c = 'B.Pharm'
subjects.append(c)
print(subjects)
```

```python
# You can append a list within a list

new = ['11th', '12th']   # Defining a new list
subjects.append(new)      # Appending new list to the old list
print(subjects)           # Note that the list itself got inserted into the list. That list constitutes one element and not two.
```

```python
print(subjects[-1])
```

```python
# Note, in the above example, the list got inserted as a list only and not as individual items.
# To add individual list items to the original list, use the function extend

help(subjects.extend)
```

```python
# The last 4 words are important in above definition - ELEMENTS FROM THE ITERABLE. ITERABLE in this case is thew new list that we are passing (see the first argument being passed to the function definiion)
# In other words, extend actually unpacks the other list and adds items to the original list instead of creating a nested list by simply inserting one list into the other

# Let's check
subjects.extend(new)
print(subjects)
```

```python
countries = ['Russia', 'India', 'US', 'Japan', 'Germany', 'Italy', 'Zambia']
```

```python
# Let's define a simple for loop

for item in countries:            # item could be changed by any other random name.
  print(item)
```

```python
for x in countries:             # x can be changed to any other word
  print(x)
```

```python
# Note that : is required at the end of for line. Also, the next code block needs to have 4 indentation (i.e. blank space).
# Check the difference between these two examples

for x in countries:
  print(x)
print('Now I am outside the for loop')

print('-------------')

for x in countries:
  print(x)
  print('Now I am inside the for loop')
```

```python
# Let's spice up the things little bit

for item in countries:
  print(item, len(item))
```

```python
for item in countries:
  print(item, item.upper(), len(item))
```

```python
for item in countries:
  print(f'{item}, {item.upper()}, {len(item)}')
```

```python
for item in countries:
  print(f'{item} --- {item.upper()} --- {len(item)}')  # f string allows to easily combine our print strings and python codes
```

```python
# Use enumerate() function to get index of element.

for id, item in enumerate(countries):   # enumerate returns index along with the actual item, hence we need to catch it explicitly.
  print(id, item)
```

```python
for index, item in enumerate(countries):   # id can be changed to anyother thing
  print(index, item)
```

```python
# let's try to create a list using for loop and append()

cubes = []    # first define a blank list
for i in range(5):       # you can explore what range(5) gives. In short, it returns numbers upto 5.
  cubes.append(i ** 3)
print(cubes)
```

```python
# To deconstruct what is exactly happening in the above code, you can use print() like this

cubes = []
for i in range(5):
  print("i = ", i)
  cubes.append(i ** 3)
  print("Number appended (i ** 3) = ", i ** 3)
  print("Current status of cubes list = ", cubes)
print()
print('Outside for loop')
print(cubes)
```
